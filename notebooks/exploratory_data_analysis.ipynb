## 4️⃣ Temporal Analysis - Time Trends

```python
# Identify relevant columns
print("📅 Time-related columns:")
time_cols = [col for col in df.columns if any(x in col.lower() for x in ['year', 'week', 'month'])]
print(time_cols)

# Check if we have dengue cases column
dengue_cols = [col for col in df.columns if 'dengue' in col.lower() or 'case' in col.lower()]
print(f"\n🦟 Dengue-related columns:")
print(dengue_cols)
```

```python
# Yearly trend of dengue cases
if 'Year' in df.columns and len(dengue_cols) > 0:
    dengue_col = dengue_cols[0]  # Use first dengue-related column
    
    yearly_cases = df.groupby('Year')[dengue_col].sum().reset_index()
    
    fig = px.line(yearly_cases, x='Year', y=dengue_col,
                  title='📈 Yearly Dengue Cases Trend (2015-2024)',
                  markers=True,
                  labels={'Year': 'Year', dengue_col: 'Total Cases'})
    
    fig.update_traces(line_color='#ff3366', marker=dict(size=10))
    fig.update_layout(height=500, hovermode='x unified')
    fig.show()
    
    print(f"\n📊 Yearly Statistics:")
    print(yearly_cases.to_string(index=False))
else:
    print("⚠️ Year or dengue cases column not found")
```

```python
# Monthly pattern analysis
if 'Month' in df.columns and len(dengue_cols) > 0:
    month_order = ['January', 'February', 'March', 'April', 'May', 'June',
                   'July', 'August', 'September', 'October', 'November', 'December']
    
    monthly_avg = df.groupby('Month')[dengue_col].mean().reset_index()
    monthly_avg['Month'] = pd.Categorical(monthly_avg['Month'], categories=month_order, ordered=True)
    monthly_avg = monthly_avg.sort_values('Month')
    
    fig = px.bar(monthly_avg, x='Month', y=dengue_col,
                 title='📊 Average Dengue Cases by Month',
                 labels={'Month': 'Month', dengue_col: 'Average Cases'},
                 color=dengue_col,
                 color_continuous_scale='Reds')
    
    fig.update_layout(height=500, xaxis_tickangle=-45)
    fig.show()
    
    print(f"\n📊 Monthly Statistics:")
    print(monthly_avg.to_string(index=False))
```

```python
# Weekly pattern within year
if 'Week' in df.columns and len(dengue_cols) > 0:
    weekly_avg = df.groupby('Week')[dengue_col].mean().reset_index()
    
    fig = px.line(weekly_avg, x='Week', y=dengue_col,
                  title='📈 Average Dengue Cases by Week of Year',
                  markers=True,
                  labels={'Week': 'Week Number', dengue_col: 'Average Cases'})
    
    fig.update_traces(line_color='#ffcc00', marker=dict(size=4))
    fig.update_layout(height=500)
    fig.show()
```

## 6️⃣ Weather Impact Analysis

```python
# Identify weather columns
weather_cols = [col for col in df.columns if any(x in col.lower() 
                for x in ['temp', 'rain', 'humid', 'weather'])]
print("🌤️ Weather-related columns:")
print(weather_cols)
```

## 9️⃣ Risk Level Analysis

```python
# Identify risk level column
risk_cols = [col for col in df.columns if 'risk' in col.lower() or 'outbreak' in col.lower()]
print("⚠️ Risk-related columns:")
print(risk_cols)
```

```python
# Risk level distribution
if risk_cols:
    risk_col = risk_cols[0]
    
    risk_dist = df[risk_col].value_counts().reset_index()
    risk_dist.columns = ['Risk Level', 'Count']
    
    # Define custom colors: High=Red, Moderate=Green, Low=Blue
    color_map = {
        0: '#00ccff',  # Low risk - Blue
        1: '#00ff88',  # Moderate risk - Green
        2: '#ff3366',  # High risk - Red
        'Low': '#00ccff',
        'Moderate': '#00ff88',
        'High': '#ff3366',
        'low': '#00ccff',
        'moderate': '#00ff88',
        'high': '#ff3366'
    }
    
    # Map colors based on risk level
    colors = [color_map.get(level, '#808080') for level in risk_dist['Risk Level']]
    
    fig = px.pie(risk_dist, values='Count', names='Risk Level',
                 title='🎯 Distribution of Risk Levels',
                 color='Risk Level',
                 color_discrete_map=color_map,
                 hole=0.4)
    
    fig.update_traces(textposition='inside', textinfo='percent+label')
    fig.update_layout(height=500)
    fig.show()
    
    print(f"\n📊 Risk Level Distribution:")
    print(risk_dist.to_string(index=False))
    print(f"\nPercentages:")
    for idx, row in risk_dist.iterrows():
        pct = (row['Count'] / risk_dist['Count'].sum()) * 100
        print(f"   {row['Risk Level']}: {pct:.2f}%")
```

```python
# Risk level by city (Top 10 cities)
if risk_cols and 'City' in df.columns:
    top_cities = df['City'].value_counts().head(15).index
    df_top_cities = df[df['City'].isin(top_cities)]
    
    city_risk = pd.crosstab(df_top_cities['City'], df_top_cities[risk_col], normalize='index') * 100
    
    # Define custom colors: High=Red, Moderate=Green, Low=Blue
    color_map = {
        0: '#00ccff',  # Low risk - Blue
        1: '#00ff88',  # Moderate risk - Green
        2: '#ff3366',  # High risk - Red
        'Low': '#00ccff',
        'Moderate': '#00ff88',
        'High': '#ff3366',
        'low': '#00ccff',
        'moderate': '#00ff88',
        'high': '#ff3366'
    }
    
    fig = go.Figure()
    
    for risk_level in city_risk.columns:
        fig.add_trace(go.Bar(
            name=str(risk_level),
            x=city_risk.index,
            y=city_risk[risk_level],
            text=[f'{val:.1f}%' for val in city_risk[risk_level]],
            textposition='inside',
            marker_color=color_map.get(risk_level, '#808080')
        ))
    
    fig.update_layout(
        title='🏙️ Risk Level Distribution by City (Top 10)',
        xaxis_title='City',
        yaxis_title='Percentage (%)',
        barmode='stack',
        height=500,
        xaxis_tickangle=-45
    )
    
    fig.show()
```

## 1️⃣1️⃣ Outlier Detection

```python
# Box plots for key numerical features
if len(numerical_cols) >= 6:
    fig = make_subplots(rows=2, cols=3, 
                        subplot_titles=numerical_cols[:6])
    
    for idx, col in enumerate(numerical_cols[:6]):
        row = idx // 3 + 1
        col_num = idx % 3 + 1
        
        fig.add_trace(go.Box(y=df[col], name=col, marker_color='lightblue'),
                     row=row, col=col_num)
    
    fig.update_layout(height=600, title_text='📦 Outlier Detection - Box Plots',
                     showlegend=False)
    fig.show()
```

```python
# Calculate IQR-based outliers
outlier_summary = []

for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    outlier_count = len(outliers)
    outlier_pct = (outlier_count / len(df)) * 100
    
    outlier_summary.append({
        'Feature': col,
        'Outliers': outlier_count,
        'Percentage': f"{outlier_pct:.2f}%",
        'Lower Bound': f"{lower_bound:.2f}",
        'Upper Bound': f"{upper_bound:.2f}"
    })

outlier_df = pd.DataFrame(outlier_summary)
outlier_df = outlier_df.sort_values('Outliers', ascending=False)

print("\n⚠️ Outlier Summary (IQR Method):")
print("="*80)
print(outlier_df.to_string(index=False))
```
